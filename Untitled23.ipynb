```python
# Step 1: Install Ultralytics (only needed once)
!pip install ultralytics --upgrade

# Step 2: Import and upload files
from ultralytics import YOLO
from google.colab import files
import matplotlib.pyplot as plt
import cv2
import os

# Upload your YOLOv11 .pt model file
print("Please upload your YOLOv11 .pt model file:")
uploaded_model = files.upload()
model_path = list(uploaded_model.keys())[0]

```

```python
# Upload an image to test
print("Now upload an image to test:")
uploaded_image = files.upload()
image_path = list(uploaded_image.keys())[0]

# Load the model
model = YOLO(model_path)  # e.g., "yolov11n.pt" or your custom "best.pt"

# Run inference
results = model(image_path)[0]

# Plot and display the result
plt.figure(figsize=(12, 8))
plt.imshow(results.plot())  # This draws boxes and labels
plt.axis('off')
plt.title(f"YOLOv11 Detection Results - {os.path.basename(image_path)}")
plt.show()

# Optional: Print detection details
print("\nDetected objects:")
for r in results:
    boxes = r.boxes
    for box in boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        label = r.names[cls]
        print(f"- {label} (confidence: {conf:.2f})")
```
